```python
lst = [1]
lst
```

```python
# append

lst = lst + [3]
lst
```

```python
# insert

def custom_list_insert(lst, index, value):
    return lst[:index] + [value] + lst[index:]

lst = custom_list_insert(lst=lst, index=1, value=2) # python indexing starts from 0
lst
```

```python
# pop

def custom_list_pop(index):
    item = lst[index]
    del lst[index]
    return item

print("Original List:", lst)

# -1 index means last item; lst is not passed, but it is modified, as lists are mutable
print("Popped item: ", custom_list_pop(index=-1)) 

print("List after pop:", lst)
```

```python
# remove

def custom_list_remove(lst, item):
    for i, val in enumerate(lst):
        if val == item:
            lst = lst[:i] + lst[i+1:]
            return

print("Original List:", lst)

print("Removing item (returns nothing):", custom_list_remove(lst, 2)) 

print("List after remove:", lst)
```

```python
# index

def custom_list_index(lst, item):
    for i, val in enumerate(lst):
        if val == item:
            return i

custom_list_index(lst=lst, item=1)            
```
